```python
#Lineplot
sns.lineplot(x='Year',y='Life_Expectancy',data=df, ci=None)    #ci=confidence interval
plt.show()
```

```python
sns.lineplot(
    x='Year',
    y='Life_Expectancy',
    hue='Country',
    data=df,
)
plt.show()
```

```python
sns.lineplot(x='Life_Expectancy',y='Spending_USD',data=df)    
plt.show()
```

```python
sns.boxplot(x='Spending_USD',data=df)
plt.show()
```

```python
#Scatter Plot 
sns.scatterplot(x='Life_Expectancy',y='Spending_USD', data=df,hue='Country')
plt.title("Scatter Plot")
plt.show()
```

```python
#barplot
sns.barplot(x='Country',
            y='Spending_USD', data=df)
plt.title("Average healthcare spending for each country")
plt.show() 
```

```python
sns.violinplot(x='Life_Expectancy',data=df)
plt.show()
```

```python
sns.histplot(x='Spending_USD', data=df, kde=True)
plt.show()
```

```python
sns.pairplot(df, hue='Country')
plt.show()
```
