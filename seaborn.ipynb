### Data Visualisation using Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# Getting the built-in datasets of Seaborn
sns.get_dataset_names()
```

```python
# Loding dataset
df=sns.load_dataset("iris")
```

```python
df
```

```python
#Regression problem ----> wherever the output is expected in countinuous data that is called as regression problem.(Salary, price)
#Binary classification ----> Only two classifictaion (true/false,eligible/not-eligible)
#Multi-class classfication ----> More than two classification (High,medium,low)
```

```python
#Lineplot
sns.lineplot(x='sepal_length',y='sepal_width',data=df, ci=None)    #ci=confidence interval
plt.show()
```

```python
#ci=confidence interval
sns.lineplot(x='sepal_length',y='sepal_width',data=df) 
plt.show()
```

```python
#lineplot using hue --->hue is used to differentiate data using different colors
sns.lineplot(x='sepal_length', y='sepal_width', data=df, hue='species',ci=None)
```

```python
#continuous and categorical feature
sns.lineplot(x='sepal_width', y='species', data=df, hue='species', ci=None)
plt.show()
```

```python
sns.lineplot(x='petal_length', y='species', data=df, hue='species', ci=None)
plt.show()
```

```python
#Scatter Plot 
sns.scatterplot(x='sepal_length',y='sepal_width', data=df, hue='species')
plt.title("Scatter Plot")
plt.show()
```

```python
sns.scatterplot(x='petal_length',y='petal_width', data=df, hue='species')
plt.title("Scatter Plot")
plt.show()
```

```python
#barplot
sns.barplot(x='species',y='sepal_length', data=df)
plt.show() 
```

```python
sns.countplot(x='species',data=df)
plt.show()
```

```python
#boxplot
sns.boxplot(x='sepal_width',data=df)
plt.show()
```

```python
sns.boxplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
sns.violinplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
sns.stripplot(x='species',y='petal_width',data=df)
plt.show()
```

```python
sns.histplot(x='petal_width',data=df)
plt.show()
```

```python
sns.histplot(x='petal_width',data=df,hue='species',kde=True)    #kde-->Kernel Density Estimation
plt.show()
#kde is appriximation of histogram to density curve.
```

```python
sns.pairplot(data=df,hue='species')
plt.show()
```

```python
#violin:
sns.catplot(x='sepal_length', data=df, kind='violin')
plt.show()
```

```python
sns.catplot(x='species', data=df, kind='violin')
plt.show()
```

```python
#swarm:
sns.catplot(x='species',y='sepal_length',data=df,kind='swarm')
plt.show()
```
